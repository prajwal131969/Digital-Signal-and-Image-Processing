```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

print("Libraries imported successfully")
```

```python
input_folder = r"c:\Users\prajw\OneDrive\Desktop\sem 5\DSIP\Open Ended Problem\LAB IMAGES"
output_folder = r"c:\Users\prajw\OneDrive\Desktop\sem 5\DSIP\Open Ended Problem\LAB IMAGES ENHANCED"
os.makedirs(output_folder, exist_ok=True)
print("Folders ready")
```

```python
image1 = cv2.imread(os.path.join(input_folder, "01.png"))
image2 = cv2.imread(os.path.join(input_folder, "02.png"))
image3 = cv2.imread(os.path.join(input_folder, "03.png"))
image4 = cv2.imread(os.path.join(input_folder, "04.png"))
image5 = cv2.imread(os.path.join(input_folder, "05.png"))

images = [image1, image2, image3, image4, image5]

for i, img in enumerate(images, 1):
    if img is None:
        print("Image", i, "not found")
    else:
        print("Image", i, "loaded successfully")
```

```python
gray1 = cv2.cvtColor(image1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(image2, cv2.COLOR_BGR2GRAY)
gray3 = cv2.cvtColor(image3, cv2.COLOR_BGR2GRAY)
gray4 = cv2.cvtColor(image4, cv2.COLOR_BGR2GRAY)
gray5 = cv2.cvtColor(image5, cv2.COLOR_BGR2GRAY)

gray_images = [gray1, gray2, gray3, gray4, gray5]

print("All images converted to grayscale")
```

```python
plt.figure(figsize=(15, 8))

for i, img in enumerate(gray_images, 1):
    plt.subplot(2, 3, i)
    plt.imshow(img, cmap="gray")
    plt.title("Image " + str(i))
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
r_min = np.min(gray1)
r_max = np.max(gray1)

contrast1 = ((gray1 - r_min) / (r_max - r_min) * 255).astype(np.uint8)

cv2.imwrite(
    os.path.join(output_folder, "image1_contrast_stretching.png"),
    contrast1
)

plt.figure(figsize=(6, 5))
plt.imshow(contrast1, cmap="gray")
plt.title("Image 1 - Contrast Stretching")
plt.axis("off")
plt.show()
```

```python
gamma = 0.6

gamma1 = 255 * (gray1.astype(np.float32) / 255) ** gamma
gamma1 = np.clip(gamma1, 0, 255).astype(np.uint8)

cv2.imwrite(
    os.path.join(output_folder, "image1_gamma.png"),
    gamma1
)

plt.figure(figsize=(6, 5))
plt.imshow(gamma1, cmap="gray")
plt.title("Image 1 - Gamma Transformation")
plt.axis("off")
plt.show()
```

```python
c = 255 / np.log(1 + np.max(gray1))

log1 = c * np.log(1 + gray1.astype(np.float32))
log1 = np.clip(log1, 0, 255).astype(np.uint8)

cv2.imwrite(
    os.path.join(output_folder, "image1_log.png"),
    log1
)

plt.figure(figsize=(6, 5))
plt.imshow(log1, cmap="gray")
plt.title("Image 1 - Log Transformation")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(gray1, cmap="gray")
plt.title("Original Image 1")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(contrast1, cmap="gray")
plt.title("Contrast Stretching")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(gamma1, cmap="gray")
plt.title("Gamma Transformation")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(log1, cmap="gray")
plt.title("Log Transformation")
plt.axis("off")

plt.tight_layout()
plt.show()
```
